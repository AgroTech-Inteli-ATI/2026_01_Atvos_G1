## Analisando a distribuicao de safras nos arquivos

```python
import pandas as pd
import os

PROCESSED = "../../data/processed"

partes = {
    "part_1": os.path.join(PROCESSED, "Inventario_atvos_21_27_part_1_silver.parquet"),
    "part_2": os.path.join(PROCESSED, "Inventario_atvos_21_27_part_2_silver.parquet"),
    "part_3": os.path.join(PROCESSED, "Inventario_atvos_21_27_part_3_silver.parquet"),
    "part_4": os.path.join(PROCESSED, "Inventario_atvos_21_27_part_4_silver.parquet"),
}

dfs = {nome: pd.read_parquet(path) for nome, path in partes.items() if os.path.exists(path)}
print(f"Partes carregadas: {list(dfs.keys())}")
```

```python

df3 = dfs["part_3"]
print(f"Shape: {df3.shape}")
print(f"\nColunas ({len(df3.columns)}):")
for col in df3.columns:
    print(f"  {repr(col):30s}  dtype={df3[col].dtype}")
```

```python

for nome, df in dfs.items():
    if "SAFRA" not in df.columns:
        cols_int = [c for c in df.columns if not isinstance(c, str)]
        print(f"{nome}: sem coluna 'SAFRA' — colunas não-string: {cols_int}")
    else:
        safras = sorted(df["SAFRA"].dropna().unique())
        print(f"{nome}: {safras}")
```

```python

rows = []
for nome, df in dfs.items():
    if "SAFRA" not in df.columns:
        continue
    for safra, n in df["SAFRA"].value_counts().sort_index().items():
        rows.append({"parte": nome, "SAFRA": safra, "linhas": n})

pd.DataFrame(rows).pivot(index="SAFRA", columns="parte", values="linhas").fillna(0).astype(int)
```
